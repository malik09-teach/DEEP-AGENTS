```python
import os
from deepagents import create_deep_agent
from langchain_community.tools.tavily_search import TavilySearchResults
from langchain_groq import ChatGroq
from dotenv import  load_dotenv
```

```python
load_dotenv()

os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGSMITH_API_KEY")
os.environ["LANGCHAIN_PROJECT"] = os.getenv("LANGSMITH_PROJECT")

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

llm = ChatGroq(model="openai/gpt-oss-120b")
```

```python
tavily_tool = TavilySearchResults(max_results=3)
```

```python
instructions = """
You are a highly capable Deep Agent. Follow these routing rules strictly:
1. If the user's query requires up-to-date facts, current events, or external data, you MUST use the Tavily search tool.
2. If the user's query is general knowledge, coding assistance, or conversational, you MUST answer directly from your own knowledge without calling the search tool.
"""
```

```python

# 5. Compile the Deep Agent 
# This one line replaces all the LangGraph StateGraph, START/END edges, and ToolNode boilerplate.
agent = create_deep_agent(
    model=llm,
    tools=[tavily_tool],
    instructions=instructions
)
```
